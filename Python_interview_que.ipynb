```python
# Write a function to find non duplicate numbers preserving order: [1,1,3,2,5,6,5] → [1,3,2,5,6]
def find_non_duplicate_no(numbers):
  result=[]
  for i in numbers:
    if i not in result:
      result.append(i)
  return result

a=find_non_duplicate_no([1,2,2,3,4,4,5])
print(a)
```
