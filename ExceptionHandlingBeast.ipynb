## try and except

```python
try:
    def division(a,b):
        return a/b
    print(division(1,1))
except:
    print('Cant divide by zero')

print('code executed')
```

```python
try:
    def division(a,b):
        return a/b
    print(division(1,0))
except Exception as e:
    print('error occured -',e)
```

```python
l=[2,0,'hello',False,None]
for i in l:
    try:
        print(f'current element ->{i}')
        result= 5/int(i)
        print(f'result- {result}')
    except Exception as e:
        print(f'exception was - {e}')
    print('*'*50)
print("Execution success")
```

```python
# finallly will always run
try:
    print("Try")
except:
    print("Error")
finally:
    print("Finally")

# why do we need finally then
# to ensure that a specific block of code is always executed
```

```python
# not that important
# else block
try:
    print("Try")
except:
    print("Error")
else:
    print("Else")
finally:
    print("Finally")
```

```python
# we are raiseing exception
# for example if possword length is less than 4 char raise errot
name=input()
try:
    if len(name)<5:
        raise Exception("Len should be > than 4")
    else:
        print("Name -",name)
except Exception as e:
    print(f'Error occured as {e}')
```

```python
# loop continuous
correct_name=False
while correct_name == False:
    name=input()
    try:
        if len(name)<5:
            raise Exception("Len should be > than 4")
        else:
            print("Name -",name)
            correct_name=True
    except Exception as e:
        print(f'Error occured as {e}')
```
