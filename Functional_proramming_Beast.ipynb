```python
def pretty(func):
  def inner():
    print("-"*50)
    func()
    print("-"*50)
  return inner
```

```python
def hello():
  print("Hello everyone")
```

```python
def bye():
  print("Bye everyone")
```

```python
hello_new=pretty(hello)
bye_new=pretty(bye)
```

```python
hello_new()
```

```python
bye_new()
```

```python
@pretty
def what():
  print('what the hell is this magic')
```

```python
'''
principles of functional programming ->

- data should be seperated from mutations
- treat variables as immutable
- treat functions as fccs or objects
'''
```

```python
x=5
# hidden mutations
x=3*x
x+=1
# hidden mutations
print(x)
# if we modify in original data, cant able to rollback to origginal data
```

```python
x=5
x1=3*x
x2=x1+1
print(x2)

#too many repetitive copies of the data mmemory issues
```

```python
# best way
x=5

def mutation_1(x):
  x=3*x
  x+=1

  return x

def mutation_1_reverse(x):
  x-=1
  x=x/3

  return x

x1=mutation_1(x)
print(x1)
```

```python
mutation_1_reverse(x1)
```

```python
def cust_sum(a,b,*args):
  result=a+b

  if args:
    result += reduce(lambda x,y:x+y,args)
  return result
```

```python
def create_person(name,age,gender,**kwargs):
  Person={
      'name':name,
      'age':age,
      'gender':gender
  }
  print("Kwargs ->", kwargs)
  return Person
```

```python
def create_person(name,age,gender,**kwargs):
  Person={
      'name':name,
      'age':age,
      'gender':gender
  }
  if kwargs:
    Person.update(kwargs)
  return Person
```

```python
def random(x,y,*args,**kwargs):
  print(x,y)
  print(args)
  print(kwargs)
```

```python
random(1,2,3,4,5,a=1,b=2)
```

```python
random(1,2,a=1,b=2,3,4,5)
```
