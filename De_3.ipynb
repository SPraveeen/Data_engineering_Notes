# Comprehensive Notes on HDFS and the Hadoop Ecosystem

## Introduction to HDFS
Hadoop Distributed File System (HDFS) stores large amounts of data across multiple machines in a reliable and scalable manner. It is a core part of the Apache Hadoop ecosystem and enables distributed storage and processing across clusters.

## Basic Structure of HDFS
- **NameNode**  
    The master service that maintains filesystem metadata and tracks where each block is stored across DataNodes.

- **DataNode**  
    Worker nodes that store actual data blocks. Each DataNode periodically reports the list of blocks it holds to the NameNode.

- **Chunking Data**  
    Files are split into blocks (chunks) to allow parallel processing and efficient storage. Typical default block sizes are 128 MB or 256 MB.

## Data Distribution and Redundancy
- **Replication**  
    HDFS ensures fault tolerance by replicating blocks across multiple DataNodes. A common replication factor is 3, meaning each block is stored on three different DataNodes.
