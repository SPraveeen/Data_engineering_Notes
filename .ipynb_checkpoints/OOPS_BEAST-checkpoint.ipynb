```python
# to stop this -> "<__main__.Car object at 0x7c799756d400>"
class Car:
  def __init__(self,name,mileage):
    self.name=name
    self.mileage=mileage

# It returns the string that gets printed when an object is printed
  def __str__(self):
    return "I am from object "

# # if we do c1+c2 it produces error to avoids this
  def __add__(self,other):
    return self.mileage + other.mileage


c1=Car("BMW",12)
c2=Car("Lambo",10)

print(c1+c2)
print()
print(c1.name)
print(c1.mileage)
print(c1)
print()
print(c2.name)
print(c2.mileage)
print(c2)
```

# Private properties

```python
class BankAccount:
  def __init__(self,balance):
    self.balance=balance

  def withdraw(self,amount):
    self.balance-=amount

  def deposit(self,amount):
    self.balance+=amount
```

```python
b=BankAccount(1000)
b.deposit(200)
print(b.balance)
# new coder writes the following line
# b.balance=876353662882029827177
b.withdraw(400)
print(b.balance)
```

```python
b=BankAccount(1000)
b.deposit(200)
print(b.balance)
# new coder writes the following line
b.balance=876353662882029827177
b.withdraw(400)
print(b.balance)
# this will ruin all bank details
```

```python
# we can restrict access (somewhat) by using private properties
```

```python
# private properties
# by adding double underscore (__) before variable
# its a private 
```

```python
class BankAccount:
  def __init__(self,balance):
    self.__balance=balance

  def withdraw(self,amount):
    self.__balance-=amount

  def deposit(self,amount):
    self.__balance+=amount

  def read_balance(self):
    return self.__balance

  def __str__(self):
      return "This is function"
```

```python
b=BankAccount(1000)
b.deposit(200)
print(b.balance)
print(b.__balance)
# this is error what we want it should nbot want to read/update directly
# new coder writes the following line
# b.balance=876353662882029827177
# b.withdraw(400)
# print(b.balance)
```

```python
# we can read by using getter function
b=BankAccount(1000)
b.deposit(200)
print(b.read_balance())
b.__balance=948475757584939
print(b.read_balance())
b.withdraw(500)
print(b.read_balance())
```

```python
# some what
b2=BankAccount(2000)
dir(b2)
# when you create a private property with __prefix__, python actually stores  it as
# __{className}__{propertyName}
```

```python
# ACTULALY WE CAN CHANGE THIS
b2=BankAccount(2000)
print(b2._BankAccount__balance)
b2._BankAccount__balance=4847738839883
print(b2.read_balance())
#some one can break this code, python is not safe really
```

```python
class SchoolMember:
    def __init__(self,name):
        self.name=name
    def introduce(self):
        return f'This is my name {self.name}'

#super method is used to call parent class
class Student(SchoolMember):
    def __init__(self,name,grade):
        super().__init__(name) #to access name instead of self.name=name
        self.grade=grade

class Staff(SchoolMember):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary=salary

class Teacher(Staff):
    def __init__(self,name,salary,subject):
        super().__init__(name,salary)
        self.subject=subject
```

```python
s1=Student("abhi","A")
print(s1.name)
print(s1.grade)
print(s1.introduce())
print()
staff1=Staff('nidhi',2000)
print(staff1.name)
print(staff1.salary)
print()
t1=Teacher("Ankita",20000,"Python")
print(t1.name)
print(t1.salary)
print(t1.subject)
```

# order of invocation
### in what order will the `init` methods complete their execution

```python
class SchoolMember:
    def __init__(self,name):
        self.name=name
    def introduce(self):
        return f'This is my name {self.name}'

#super method is used to call parent class
class Staff(SchoolMember):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary=salary

class Teacher(Staff):
    def __init__(self,name,salary,subject):
        super().__init__(name,salary)
        self.subject=subject

t=Teacher('tarun',5,'python')
print(Teacher.__mro__) #method resolution order
```

## Multiple Inheritance

```python
# when class has two parents
# Diamond problem
class A:
    x=10

class B(A):
    pass

class C(B):
    pass

class D(A):
    x=5

class E(C,D):
    pass

e=E()
print(e.x)
```

```python
print(E.__mro__)
```
