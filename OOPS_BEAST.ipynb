```python
# to stop this -> "<__main__.Car object at 0x7c799756d400>"
class Car:
  def __init__(self,name,mileage):
    self.name=name
    self.mileage=mileage

# It returns the string that gets printed when an object is printed
  def __str__(self):
    return "I am from object "

# # if we do c1+c2 it produces error to avoids this
  def __add__(self,other):
    return self.mileage + other.mileage


c1=Car("BMW",12)
c2=Car("Lambo",10)

print(c1+c2)
print()
print(c1.name)
print(c1.mileage)
print(c1)
print()
print(c2.name)
print(c2.mileage)
print(c2)
```

# Private properties

```python
class BankAccount:
  def __init__(self,balance):
    self.balance=balance

  def withdraw(self,amount):
    self.balance-=amount

  def deposit(self,amount):
    self.balance+=amount
```

```python
b=BankAccount(1000)
b.deposit(200)
print(b.balance)

# new coder writes the following line
b.balance=876353662882029827177

b.withdraw(400)
print(b.balance)
```
